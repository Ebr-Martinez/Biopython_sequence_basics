# Basic analysis using Biopython

This script include basic sequence analysis using some Biopython tools for sequence manipulation and analysis

```python
from Bio.Seq import Seq

seq1 = Seq("AGTGGTGTAACCACAGGTGGACTA")
```

```python
seq1.complement()
```

```python
seq1.reverse_complement()
```

```python
seq1.translate()
```

```python
len(seq1)
```

```python
for nucleotide in ["A", "C", "G", "T"]:
    print(nucleotide, seq1.count(nucleotide))
```
